# Bronze - Countries Data

## Setup Environment and Prepare data 

```python
from letterboxd_data_pipeline.load_data import load_files_from_layer

bronze_df = load_files_from_layer(layer="bronze", file_list=["countries"])
bronze_countries_df = bronze_df["countries"]
```

## Movie with most countries

```python
bronze_countries_df.groupby("id").agg(total_countries=("country", "count")).sort_values(by="total_countries",ascending=False)
```
