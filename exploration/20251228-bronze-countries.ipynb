# Bronze - Countries Data

## Setup Environment and Prepare data 

```python
from letterboxd_data_pipeline.load_data import load_files_from_layer
from letterboxd_data_pipeline.explore_data import is_possible_na

bronze_df = load_files_from_layer(layer="bronze", file_list=["countries.parquet"])
bronze_countries_df = bronze_df["countries"]
```

## Simple Explore

```python
bronze_countries_df.describe(include="all")
```

## Number of NA

```python
bronze_countries_df.isna().sum()
```

## Possible NA

```python
possible_na_df = bronze_countries_df[is_possible_na(bronze_countries_df["country"])]
possible_na_df
```

## Movie with most countries

```python
bronze_countries_df.groupby("id").agg(total_countries=("country", "count")).sort_values(by="total_countries",ascending=False)
```
