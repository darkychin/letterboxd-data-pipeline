# Bronze - Curious Data

Curious data and probable easter eggs that is worth mentioning.

## Setup Environment and Prepare data 

```python
from letterboxd_data_pipeline.load_data import load_files_from_layer
from letterboxd_data_pipeline.explore_data.explore_data import is_possible_na

bronze_df = load_files_from_layer(
    layer="bronze",
    file_list=[
        "actors.parquet",
        "movies.parquet",
        "countries.parquet",
        "crew.parquet",
        "releases.parquet",
    ],
)
actors_df = bronze_df["actors"].copy(True)
movies_df = bronze_df["movies"].copy(True)
countries_df = bronze_df["countries"].copy(True)
crews_df = bronze_df["crew"].copy(True)
releases_df = bronze_df["releases"]
```

## Curious Actor and Movie Details
