# Bronze - Crew - Column Characters Frequency Explore

```python
import pandas as pd
import plotly.express as px
from letterboxd_data_pipeline.load_data import load_files_from_layer
from letterboxd_data_pipeline.explore_data.char_frequency import gen_char_frequency_df

bronze_df = load_files_from_layer(layer="bronze", file_list=["crew.parquet"])
target_df = bronze_df["crew"]
```

```python
target_df.info()
```

## Column "name"

```python
df = gen_char_frequency_df(target_df["name"])
df_head = df.head(100)
df_tail = df.tail(1000)
```

```python
fig = px.bar(x=df_head["character"], y =df_head["frequency"])
fig.show()
```

```python
fig_tail = px.bar(x=df_tail["character"], y =df_tail["frequency"])
fig_tail.show()
```

```python
# pattern = r"[^\p{L}\p{M}\s\-\']+" // required library "regex"
pattern = r"[^\w\x00-\x7F]"

special_char_df = df[df["character"].str.contains(pattern, regex=True)]
special_char_df.info()
```

```python
fig_symbol = px.bar(x=special_char_df["character"], y=special_char_df["frequency"])
fig_symbol.show()
```
