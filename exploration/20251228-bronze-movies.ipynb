## Possible NA

```python
possible_na_df = bronze_movies_df[
  is_possible_na(bronze_movies_df["name"]) | 
  is_possible_na(bronze_movies_df["tagline"]) |
  is_possible_na(bronze_movies_df["description"])
]
possible_na_df
```

## Most Popular Movie Name

```python
# Always use this line to sample the entire data to verify the code swiftly
popular_name_df = bronze_movies_df.copy(True)

popular_name_df.groupby("name").agg(total_count=("id","count")).sort_values(by="total_count", ascending=False)
```
