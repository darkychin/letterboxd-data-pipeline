# Bronze - Genres Data

## Setup Environment and Prepare data 

```python
from letterboxd_data_pipeline.load_data import load_files_from_layer

bronze_df = load_files_from_layer(layer="bronze", file_list=["genres.parquet"])
bronze_genres_df = bronze_df["genres"]
```

## Simple Explore

```python
bronze_genres_df.describe(include="all")
```

```python
bronze_genres_df.info()
```

## Number of NA

```python
bronze_genres_df.isna().sum()
```

## Top 10 Genre

```python
bronze_genres_df.groupby("genre").agg(total=("id","count")).sort_values("total", ascending=False).head(10)
```

## Movie with most Genre

```python
most_flavored_movies_df = bronze_genres_df.copy(True)

grouped_source_df = most_flavored_movies_df.groupby(by=["id"])

most_flavored_movies_df["genre_count"] = grouped_source_df["id"].transform(len)

most_flavored_movies_df = most_flavored_movies_df[
    ["id", "genre", "genre_count"]
].sort_values(["genre_count", "genre"], ascending=[False, True])

most_flavored_movies_df
```
