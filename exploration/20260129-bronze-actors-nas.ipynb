# Brone - Actors - Possible NAs

```python
from letterboxd_data_pipeline.load_data import load_files_from_layer
from letterboxd_data_pipeline.explore_data.explore_data import is_possible_na
import numpy as np

bronze_df = load_files_from_layer(layer="bronze", file_list=["actors.parquet"])["actors"]
bronze_df.info()
```

```python
nas_df = bronze_df[bronze_df["name"].isna() | bronze_df["role"].isna()]
nas_df
```

```python
possible_nas_df = bronze_df[is_possible_na(bronze_df["name"]) | is_possible_na(bronze_df["role"])]
possible_nas_df
```

```python
role_nas_df = bronze_df[is_possible_na(bronze_df["role"])]
role_nas_df.groupby("role").agg(total=("id","count"))
```

```python
name_nas_df = bronze_df[is_possible_na(bronze_df["name"])]
name_nas_df.groupby("name").agg(total=("id","count"))
```

Conclusion for this exploration is that both name and role can not confidently treated as `None`.
