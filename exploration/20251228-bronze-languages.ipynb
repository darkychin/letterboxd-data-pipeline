### Movie with Most Language of all Type

```python
movie_with_most_lang_df = bronze_language_df.copy(True)

grouped_source_df = movie_with_most_lang_df.groupby(by=["id", "type"])

movie_with_most_lang_df["language_count"] = grouped_source_df["language"].transform(len)

movie_with_most_lang_df = movie_with_most_lang_df[
    ["id", "type", "language", "language_count"]
].sort_values(["language_count"], ascending=[False])

movie_with_most_lang_df
```
