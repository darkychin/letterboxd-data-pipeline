# Bronze - Themes Data

## Setup Environment and Prepare data 

```python
import config
from letterboxd_data_pipeline.load_data import load_files_from_layer

bronze_df = load_files_from_layer(layer="bronze", file_list=["themes"])
bronze_themes_df = bronze_df["themes"]
```

## Simple Explore

```python
bronze_themes_df.describe(include="all")
```

## Most popular theme

```python
bronze_themes_df.groupby("theme").agg(count=("id","count")).sort_values(by="count", ascending=False)
```
