## Movie with Most Studios

```python
most_studios_df = bronze_studios_df.copy(True)

most_studios_df = most_studios_df.drop_duplicates()

most_studios_df["studio_count"] = (
    most_studios_df.groupby(by=["id"])["studio"]
    .transform(lambda x: x.nunique())
)

most_studios_df.sort_values(["studio_count", "id"], ascending=[False, True])
```
